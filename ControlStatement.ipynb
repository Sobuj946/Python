```python
n=int(input())
target=float(input())
sum=0.0
for i in range(0,n):
    loss=float(input())
    sum=sum+loss
average=sum/n
if target>=average:
    print("PASS")
else:
    print("RETRY")
```

```python
n=int(input())
yes_count=0;no_count=0
for i in range(0,n):
    inp=input()
    if inp in "YES":
        yes_count=yes_count+1
    elif inp in "NO":
        no_count=no_count+1
print(yes_count,no_count)
if yes_count>=no_count:
    print("ACCEPT")
else:
    print("REJECT")
```

```python
n=int(input())
lists=input().split()
array=[]
for i in range(0,len(lists)):
    array.append(int(lists[i]))
lowest=min(array)
mx=max(array)
for i in range(0,len(array)):
    if array[i]==mx:
        array[i]=lowest
    elif array[i]==lowest:
        array[i]=mx
print(array)
```

```python
lists=input().split()
array=[]
for i in range(0,len(lists)):
    array.append(int(lists[i]))
sum=0
for i in array:
    sum=sum+i
# print(sum)
result=sum//len(array)
if result<85:
    print("Dark Image")
elif 85<=result and result<=180:
    print("Normal Image")
elif result>170:
    print("Bright Image")
```
