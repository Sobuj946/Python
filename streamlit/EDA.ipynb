# EDA->Exploratory Data Analysis 
EDA helps uncover hidden patterns, detect outliers, test assumptions, and verify data quality, ensuring that models are built on reliable and interpretable data

 It involves investigating a dataset using statistical summaries, visualizations, and pattern recognition techniques to understand its structure, relationships, and anomalies before applying any ML algorithms


 
 # Objectives of EDA in ML
1.Understand variable distributions: Identify skewness, central tendency, and spread of numerical features.
2.Detect relationships: Examine correlations and interactions between features and the target variable.
3.Identify anomalies: Spot missing values, duplicates, or inconsistent entries that could affect model performance.
4.Feature selection: Highlight important variables that contribute most to predictive modeling.
5.Guide preprocessing: Inform decisions on scaling, encoding, or transforming features for ML algorithms
# Key Steps in EDA

Univariate Analysis: Study one variable at a time using histograms, box plots, or bar charts to understand distributions and detect outliers 

Bivariate Analysis: Explore relationships between two variables using scatter plots, correlation coefficients, cross-tabulations, or line graphs 

Multivariate Analysis: Examine interactions among multiple variables to detect multicollinearity or complex patterns that may influence model performance 

Handling Missing Values: Identify missing data and decide whether to remove or impute values to avoid bias 

Outlier Detection: Use techniques like the Interquartile Range (IQR) to detect and handle extreme values 

Feature Transformation and Scaling: Apply log, square root, or Box-Cox transformations for skewed data and standardize or normalize features for algorithms sensitive to scale 

Encoding Categorical Variables: Convert categorical data into numeric form using one-hot, label, or frequency encoding for ML models
