# Apply function

```python
#Min max scalling
mn=df['Total summation'].min()
mx=df['Total summation'].max()

df['Scaled Marks']=df['Total summation'].apply(lambda x:(x-mn)/(mx-mn))
df
```

```python
#custom build function
def grading_system(marks):
    if marks>=260:
        return 'A+'
    elif marks>=250:
        return 'A'
    else:
        return 'A-'
df['Grade']=df['Total summation'].apply(grading_system)
```

```python
df
```

```python
def marking_system(df):
    a=df['Data Structure Marks']*2
    b=df['Python Marks']*2
    c=df['Algorithm Marks']*4
    return a+b+c

df['Exception Marks']=df.apply(marking_system,axis=1)
```
