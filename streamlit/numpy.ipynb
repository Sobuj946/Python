# Array Manipulation

```python
arr1=np.random.randint(1,10,size=(4,5))
# print(arr1)

#changing the shape of the array

arr1=arr1.reshape(2,10)
# print(arr1)

arr1.ndim
# changing the dimetion of the array
flatten=arr1.flatten()
flatten
print(flatten.ndim)
```

```python
# Column wise flattening 

print(arr1)

column_wise_flattening=np.ravel(arr1,order='F')#column wise flattening


print(column_wise_flattening)

row_wise_flattening=np.ravel(arr1,order='C')#row wise flattening

print(row_wise_flattening)
```

# Concatenate

```python
a=np.random.randint(1,10,size=(2,3))
b=np.random.randint(20,30,size=(2,3))

print(a)
print(b)

concatenate_row_wise=np.concatenate((a,b),axis=0) # row wise concatenate

concatenate_column_wise=np.concatenate((a,b),axis=1) # column wise concatenate

print(concatenate_column_wise)

print(concatenate_row_wise)
```

# Arithmetic operator and mathemetical operation

```python
# These operations are performed through vectorization

x=np.array([1,23,43,4,5,6])
y=np.array([3,4,5,67,7,7])

#Element wise addition 

add=x+y
print(add)

#Element wise subtraction
 
sub=x-y
print(sub)

#Element wise multiplication
#  
mul=x*y
print(mul)

#Element wise division
 
div=x/y
print(div)

#Reminder of the array

rem=x%y
print(rem)
```

# Mathemetical operations

```python
sine_value=np.sin(x)
#sin,cos,tan these are solved through the redian
print(sine_value)

#redian to degree convert

deg_convert=np.rad2deg(x)
print(deg_convert)

#degree to redian convert
redian_convert=np.deg2rad(deg_convert)

print(redian_convert)
```

```python
#2D matrix sort

array=np.random.randint(1,20,size=(4,5))

print("Before sorting the array:")

print(array)

#Sorting the value of horizontal
#axis=1

horizontal=np.sort(array,axis=1)
print("After sorting \n")
print(horizontal)


#Sorting the value of vartical
#axis=0
vartical=np.sort(array,axis=0)
print("After vartical sorting :\n")
print(vartical)
```

## counting with ndarray

```python
array=np.random.randint(1,10,10)

print(array)

counting=np.count_nonzero(array<5)#counting the value

print(counting)

#unique  value

unique=np.unique(array)

print("unique  value:",unique)

# return counts value 

counts=np.unique(array,return_counts=True)

print(counts) #value and frequence duitai dicche
```

```python
#statistical

study_hours=np.array([2,4,6,7,8])
exam_scores=np.array([65,75,78,88,92])

data=np.array([study_hours,exam_scores])
correlation=np.corrcoef(data)

print("Correlation matrix:\n",correlation)
```

## linear algebra

```python
A=np.array([[2,3,43],[3,4,5]])
B=np.array([[5,6,7],[3,4,6],[4,5,6]])

dot_product=np.dot(A,B)

print(dot_product)


#matrix trace

print(np.trace(B))
```

```python
#matrix det

det=np.linalg.det(B)

print(det)
```
