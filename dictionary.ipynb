Generator

```python
lst=[x for x in range(500)]
# print(lst)
def provid_generator(chunk_size,lst):
    for i in range(0,len(lst),chunk_size):
        yield lst[i:i+chunk_size]
x=provid_generator(5,lst)
# print(type(x))
print(next(x))
print(next(x))
print(next(x))
```

Filter function

```python
numbers=[x for x in range(100)]

even=list(filter(lambda x:x%2==0,numbers))
print(even)

fifty_upper=list(filter(lambda x:x>50,numbers))
print(fifty_upper)
```

```python
lst=[None,"hello",2,2,3,4,None,False,True]

newlst=list(filter(None,lst))
print(newlst)
```

```python
string=['aaa','bbb','cccc','a','b','bb','aaaa','c','d']

filterString=list(filter(lambda x:len(x)>2,string))
print(filterString)
```
