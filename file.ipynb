```python
from functools import reduce
with open("sample.txt",'r') as file:
    content=file.readlines()
    total_line=len(content)
    print(len(content))
    number_of_word=list(map(lambda x:len(x.split()),content))
    print(number_of_word)
    total_number_of_word=reduce(lambda x,y:x+y,number_of_word)
    print(total_number_of_word)
    
    #cleaning process
    content=list(map(str.strip,content))#delete space
    print(content)
    #delete space
    content=list(map(lambda x:x.replace(" ",""),content))
    print(content)

    #Total character
    number_of_character=list(map(lambda x:len(x),content))
    print(number_of_character)

    total_number_of_character=reduce(lambda x,y:x+y,number_of_character)
  
    print(total_number_of_character)

    with open("count_total.txt",'w') as file:
        file.write(f"Total line :{total_line} \n Total Number of word : {total_number_of_word} \n Total Number of character : {total_number_of_character}")
```

```python
##Base class create
class Phone:
    category="Electronics"
    #constructor
    def __init__(self,model,camera,battery,battery_parcentage=100):
        self.model=model
        self.camera=camera
        self.battery=battery
        self.battery_parcentage=battery_parcentage
    def charge(self,hour,):
        self.battery_parcentage+=hour
    def capture(self,photo):
        if(self.battery_parcentage==0):
            print("Cannot click the photo")
        else:   
            self.battery_parcentage-=photo
            print("Click the photo")
            print(self.battery_parcentage)

```

```python
#Object create

iphone=Phone("iphone",45,5000)
symphony=Phone("I-34",30,400)
print(symphony.camera)
print(iphone.camera)
iphone.capture(10)
print(iphone.battery_parcentage)
```

```python
##Derived class
class SmartPhone(Phone):
   def __init__(self,model,camera,battery,processor):
        super().__init__(model,camera,battery)
        self.processor=processor
```

```python
pro=SmartPhone("x",34,5000,"snapdragon")
print(pro.processor)
pro.capture(59)
```

Problem 4: Square of Even Numbers (Map + Filter + Lambda)
Problem:
 Write a one-line Python expression using map(), filter(), and lambda that takes a list of integers and returns the squares of even numbers only

```python
nums = [1, 2, 3, 4, 5, 6]
square=list(map(lambda x:x**2 if x%2==0 else None,nums))
square=list(filter(lambda x :x is not None,square))
print(square)
```

```python
import speech_recognition as sr
import pyttsx3
import datetime
import pywhatkit
import wikipedia
import pyjokes
listener=sr.Recognizer()
alexa =pyttsx3.init()
voices=alexa.getProperty('voices')
alexa.setProperty('voice', voices[1].id)
def talk(text):
    alexa.say(text)
    alexa.runAndWait()
def take_command():
    try:
        with sr.Microphone() as source:
            print("Listening...")
            voice=listener.listen(source)
            command=listener.recognize_google(voice)
            command=command.lower()
            if 'alexa' in command:
                command=command.replace('alexa','')
    except :
        pass
    return command
def run_alexa():
    command=take_command()
    if 'time' in command:
        time=datetime.datetime.now().strftime('%I:%M %p')
        print(time)
        talk('current time is'+time)
    elif 'play' in command:
        song=command.replace('play','')
        talk('playing'+song)
        pywhatkit.playonyt(song)
    elif 'tell me about' in command:
        look_for=command.replace('tell me about','')
        info=wikipedia.summary(look_for,sentences=3)
        print(info)
        talk(info)
    elif 'jokes' in command:
        # joke=command.replace('jokes','')
        talk(pyjokes.get_jokes())
    elif 'date' in command:
        talk('Sorry vaiya , I am in another relation')
    else:
        talk('Please say again')
        pywhatkit.search(command)
while True:
    run_alexa()
```
